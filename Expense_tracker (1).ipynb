```python
import mysql.connector
import datetime


# Create a connection to the database
def conn():
    return mysql.connector.connect(
        host="localhost",
        user="root",
        password="Surya@9373",
        database="expense_tracker"
    )


# Add a new expense
def add_expense():
    category = input("Enter category like medicines, grocery: ")
    amount = float(input("Enter amount spent: "))
    date_str = input("Enter date in format yyyy-mm-dd (press Enter for today): ")
    description = input("Enter description (optional): ")

    # Use today's date if user does not enter a date
    date = date_str if date_str else datetime.datetime.today().strftime("%Y-%m-%d")

    sql = """
        INSERT INTO expense (expense_dt, category, amount, descript)
        VALUES (%s, %s, %s, %s)
    """

    vals = (date, category, amount, description)

    con = conn()
    cursor = con.cursor()

    cursor.execute(sql, vals)
    con.commit()

    print("Expense saved successfully!")

    cursor.close()
    con.close()


# View all expenses
def view_all():
    con = conn()
    cursor = con.cursor()

    cursor.execute("SELECT * FROM expense")

    records = cursor.fetchall()

    print("\nID | Date | Category | Amount | Description")
    print("-" * 60)

    for row in records:
        print(f"{row[0]} | {row[1]} | {row[2]} | {row[3]} | {row[4]}")

    cursor.close()
    con.close()


# Delete an expense
def delete_expense():
    sr_no = int(input("Enter ID or Sr_no of expense to be deleted: "))

    con = conn()
    cursor = con.cursor()

    sql = "DELETE FROM expense WHERE sr_no = %s"

    cursor.execute(sql, (sr_no,))
    con.commit()

    if cursor.rowcount > 0:
        print("Expense deleted successfully!")
    else:
        print("Expense ID not found.")

    cursor.close()
    con.close()


# Monthly expense summary
def summary():
    month = int(input("Enter month as number between 1 and 12: "))
    year = int(input("Enter year in format yyyy: "))

    if month < 1 or month > 12:
        print("Invalid month!")
        return

    con = conn()
    cursor = con.cursor()

    sql = """
        SELECT category, SUM(amount)
        FROM expense
        WHERE MONTH(expense_dt) = %s
        AND YEAR(expense_dt) = %s
        GROUP BY category
    """

    cursor.execute(sql, (month, year))

    records = cursor.fetchall()

    print("\nCategory | Total Spent")
    print("-" * 30)

    for row in records:
        print(f"{row[0]} | {row[1]}")

    cursor.close()
    con.close()


# Main function
def main():

    while True:

        print("\n---- Track Your Expenses ----")
        print("1. Add a new expense")
        print("2. Delete an expense")
        print("3. View all expenses")
        print("4. View monthly summary")
        print("5. Exit")

        ch = int(input("Enter your choice (1-5): "))

        if ch == 1:
            add_expense()

        elif ch == 2:
            delete_expense()

        elif ch == 3:
            view_all()

        elif ch == 4:
            summary()

        elif ch == 5:
            print("Thank you for using Expense Tracker!")
            break

        else:
            print("Wrong choice! Please retry.")


# Start the program
if __name__ == "__main__":
    main()
```
